```python
import plotly.express as px
```

```python
df = px.data.iris()
df
```

```python
fig  = px.histogram(df, x= 'petal_length')
fig
```

```python
fig  = px.histogram(df, x= 'petal_length', color = 'species', barmode = 'overlay')
fig
```

```python
fig  = px.histogram(df, x= 'petal_length', color = 'species', facet_col = 'species')
fig
```
