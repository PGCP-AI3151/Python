1.	Draw a bar chart showing GDP percentage for European countires. [use gdpminder dataset]
	Use different semantics as orientation, bar mode, bar norm, text to improve your chart

```python
import plotly.express as px
```

```python
df = px.data.gapminder()
df.head()
```

```python
filt = df['continent'] == 'Europe'
df_f = df[filt]
df_f
```

```python
fig = px.bar(df_f, x='country', y = 'gdpPercap', animation_frame = 'year')
fig.update_layout(
    title = 'GDP Precap Over The Year',
    xaxis_title = 'Country',
    yaxis_title = 'GDP Precap',
    barmode='stack'
  )
fig.update_traces(
    text=df['gdpPercap'],
    texttemplate='$%{text:,.0f}' # Formats as currency with commas, no decimals
)

fig
```
