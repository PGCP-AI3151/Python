```python
import plotly.express as px
```

```python
dir(px.data)
```

```python
df = px.data.gapminder()
df
```

```python
flit = df['continent'] == 'Asia'
df_f = df.loc[flit]
```

```python
fig  = px.bar(df_f, x='country', y='lifeExp')
fig
```

```python
fig  = px.bar(df, x='country', y='lifeExp', animation_frame = 'year')
fig
```

```python
fig  = px.bar(df_f, x='country', y='lifeExp', color ='country', animation_frame = 'year')
fig.update_layout(title = 'Life Exp Over The Years',
                  xaxis_title = 'Country',
                  yaxis_title = 'Life Expectancy'
                 )
fig
```
