-----Q2-----

```python
df = pd.read_csv('solar.csv')
df
```

a. Display all the plants with capacity > 500 MW 

```python
filt = df['Installed Capacity (MW)'] > 500
df[filt]
```

b. Display plant details for New York plant

```python
df[df['State'] == 'New York']
```

 c. For all the plants display Average MW Per Plant and Generation (GWh)

```python
df[['Average MW Per Plant','Generation (GWh)']]
```

 d. Sort the plants by Generation (GWh) (ascending) and Installed Capacity (MW)(descending) 

```python
print('---By Generation (GWh) In Ascending---')
df.sort_values(by='Generation (GWh)')
```

```python
print('---By Installed Capacity (MW) In Deascending---')
df.sort_values(by='Installed Capacity (MW)' , ascending = False)
```

e. List top 5 plants for their Generation

```python
print('----- Top 5 In Generation -----')
df.sort_values(by='Generation (GWh)' , ascending = False).head(5)
```

f. Display all plant details for states – California, Nevada, Arizona and Texas

```python
df[df['State'].isin(['California','Nevada','Arizona','Texas'])]
```

 g. Group the plants by region, find – min/max capacity plant for each group

```python
df.groupby('Region')['Installed Capacity (MW)'].agg(Min_Capacity = min , Max_Capacity = max).reset_index()
```

 h. Add following information to the existing details for the plants North Carolina 148 669 4.5 1162 North

```python
new_data = {'State':'North Carolina',
            'Number of Solar Plants' :148,
            'Installed Capacity (MW)' :669,
            'Average MW Per Plant' :4.5,
            'Generation (GWh)' :1162,
            'Region' : 'North'
           }
df = pd.concat([df,pd.DataFrame([new_data])],ignore_index = True)
df
```

------ Q3 ------

```python
df = pd.read_csv('gapminder2007.csv')
df
```

a.	Display data for all Asian countries

```python
df[df['continent'] == 'Asia']
```

b.	Display top and bottom 15 rows

```python
print('------ Top 15 ------')
df.head(15)
```

```python
print('------ Bottom 15 ------')
df.tail(15)
```

c.	Show all the rows where life expectancy is more than 50 years and less than 80 years

```python
df[(df['lifeExp'] > 50) & (df['lifeExp'] < 80)]
```

d.	Show data for India, America and France

```python
df[df['country'].isin(['India','America','France'])]
```

e.	Show data for all countries where name starts with ‘A’

```python
filt = list(map(lambda x : x.startswith('A'),df['country']))
df[filt]
```

f.	Show min/max and mean GDP and Population for each continent

```python
df.groupby('continent').agg(Min_GDP =('gdpPercap','min'), Max_GDP = ('gdpPercap','max'), Mean_GDP = ('gdpPercap','mean'),Min_Pop =('pop','min'), Max_Pop = ('pop','max'), Mean_Pop = ('pop','mean')).reset_index()
```

g.	Show population and life expectancy for all countries 

```python
df[['country','pop','lifeExp']]
```

h.	Sort the data by country names (ascending) and GDP (descending)

```python
print('----- Sort By Country Name -----')
df.sort_values(by = 'country' ,ascending = True)
```

```python
print('----- Sort By GDP Name -----')
df.sort_values(by = 'gdpPercap' ,ascending = False)
```

i.	Find top 20 most populous countries

```python
print('----- Top 20 Most Populous Countries -----')
df.sort_values(by = 'pop',ascending = False).head(20)
```

j.	Delete all rows for a specific country

```python
df.drop(df[(df['country'] == 'India')].index,inplace = True)
df
```

```python
df[df['country'] == 'India']
```

k.	Change all column names to title case

```python
df.columns = list(map(lambda x : x.title(),df.columns))
df
```

l.	Change column names: pop to Population, gdpPercap to GDP_Percap

```python
df.rename (columns = {'Pop':'Population','Gdppercap':'GDP_Percap'},inplace = True)
df
```

m.	Increase Life Expectancy by 2 years for all African countries

```python
df['Lifeexp'] = [x+2 for x in df['Lifeexp']]
df
```
