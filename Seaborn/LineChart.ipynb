```python
import seaborn as sns
```

```python
df =  sns.load_dataset('flights')
df
```

```python
sns.lineplot(df,x='month', y ='passengers')
```

```python
sns.lineplot(df,x='month', y ='passengers',errorbar = ('ci',80))
```

```python
sns.lineplot(df,x='year', y ='passengers')
```

```python
sns.lineplot(df,x='year', y ='passengers', hue='month')
```

```python
sns.lineplot(df,x='year', y ='passengers', hue='month',size ='month')
```

```python
sns.lineplot(df,x='year', y ='passengers', hue='month',style ='month')
```
