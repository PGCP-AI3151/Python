```python
import seaborn as sns
import matplotlib.pyplot as plt
```

------ Q1 ------

1.	Draw a histogram showing distribution of flipper_length across species for penguin dataset. Create a stack plot and step plot- – Write down your interpretations

```python
df = sns.load_dataset('penguins')
df.head()
```

----- Step Plot -----

```python
sns.histplot(df, x = 'flipper_length_mm', hue = 'species',element = 'step')
plt.title('Distibution Of Flipper Length Across Species', fontweight = 'bold')
plt.xlabel('Flipper Length (MM)', fontweight = 'bold')
plt.ylabel('Count', fontweight = 'bold')
plt.show()
```

Interpretation -> 

----- Stack Plot -----

```python
sns.histplot(df, x = 'flipper_length_mm', hue = 'species', multiple = 'stack')
plt.title('Distibution Of Flipper Length Across Species', fontweight = 'bold')
plt.xlabel('Flipper Length (MM)', fontweight = 'bold')
plt.ylabel('Count', fontweight = 'bold')
plt.show()
```

2.	Draw a bivariant KDE plot to show relationship between horsepower and displacement [use mpg dataset] – Write down your 	interpretations 

```python
df = sns.load_dataset('mpg')
df.head()
```

```python
sns.kdeplot(df, x = 'horsepower', y= 'displacement', hue = 'cylinders')
plt.title('HorsePower vs Displacement Based On No of Cylinders', fontweight = 'bold')
plt.xlabel('Horse Power', fontweight = 'bold')
plt.ylabel('Displacement', fontweight = 'bold')
plt.show()
```

3.	Draw a correlation heatmap for diamonds dataset, add annotation, change colour pallet and write your interpretation 

```python
df = sns.load_dataset('diamonds')
df.head()
```

```python
df = df.drop(['cut', 'color', 'clarity'],axis = 1)
df.head()
```

```python
sns.heatmap(df.corr(),annot = True, cmap = 'RdPu', fmt = '.1f', linewidth = 0.5)
plt.title('Correlation HeatMap', fontweight = 'bold')
plt.show()
```
