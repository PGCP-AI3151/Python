```python
import seaborn as sns
```

```python
df = sns.load_dataset('mpg')
df
```

```python
df = df.loc[:,'mpg':'acceleration']
df
```

```python
df.corr()
```

```python
sns.heatmap(df.corr())
```

```python
sns.heatmap(df.corr(),cmap='Greys')
```

```python
sns.heatmap(df.corr(),cmap='RdPu')
```

```python
sns.heatmap(df.corr(),cmap='RdPu',annot=True)
```

```python
sns.heatmap(df.corr(),cmap='RdPu',annot=True,fmt='.1f',linewidth=0.5)
```

```python
sns.heatmap(df.corr(),cmap='RdPu',annot=True,fmt='.1f',linewidth=0.7,linecolor='black')
```
