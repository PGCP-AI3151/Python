```python
import pandas as pd
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('iris')
df
```

```python
sns.histplot(df,x = 'petal_length',hue = 'species')
```

```python
sns.histplot(df,x = 'petal_length',hue = 'species', multiple = 'stack')
```

```python
sns.histplot(df,x = 'petal_length',hue = 'species', element = 'step')
```

```python
sns.displot(df,x = 'petal_length',hue = 'species', col = 'species')
```

```python
sns.displot(df,x = 'petal_length',hue = 'species', row = 'species')
```
