```python
import seaborn as sns
```

```python
df = sns.load_dataset('diamonds')
df
```

```python
sns.scatterplot(df, x='carat',y='price')
```

```python
df_f = df.sample(100)
df_f
```

```python
sns.scatterplot(df_f, x='carat',y='price')
```

```python
sns.scatterplot(df_f, x='carat',y='price',hue='cut')
```

```python
sns.scatterplot(df_f, x='carat',y='price',hue='cut',size='cut')
```

```python
sns.scatterplot(df_f, x='carat',y='price',hue='cut',style='cut')
```

```python
sns.scatterplot(df_f, x='carat',y='price',hue='cut',size='cut',alpha=0.5)
```
