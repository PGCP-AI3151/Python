The main difference between a copy and a view in a NumPy array is data ownership and memory allocation.

```python
import numpy as np
```

```python
arr = np.array([1,2,7,8,5])
arr
```

copy() -> to make copy on array , don't affect any changes happend in original
-> A copy creates an entirely new array with its own dedicated memory buffer,

```python
x = arr.copy()
x
```

arr[0] = 50   // no reflact copy array when use copy()
arr

```python
x
```

view() -> a view shares the same memory buffer as the original array. any change in original can affect view.

```python
y = arr.view()
y
```

```python
arr[0] = 50
arr
```

```python
y
```

```python
y[1] = 60
y
```

```python
arr
```
