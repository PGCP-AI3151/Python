```python
import numpy as np
```

• Normalize a NumPy array [ to normalize means scaling the values of an array into a specific range, often [0, 1]]

```python
arr = np.random.randint(1,30,9).reshape(3,3)
arr
```

```python
normalized = (arr - np.min(arr))/(np.max(arr) - np.min(arr))
normalized
```

• Create a 5×5 array with 1 on border and 0 inside

```python
arr = np.ones((5,5),dtype=int)
arr
```

```python
arr [1:4,1:4] = 0
arr
```

• Create a random array of 10×10: perform stats (min, max, mean, std)

```python
arr = np.random.randint(1,500,100).reshape(10,10)
arr
```

```python
print(np.min(arr))
```

```python
print(np.max(arr))
```

```python
print(np.mean(arr))
```

```python
print(f'{np.std(arr):.2f}')
```
