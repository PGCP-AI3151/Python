• Find common elements between two arrays [with and without using intersect1d, use isin() instead]

```python
arr1 = np.random.randint(1,10,4).reshape(2,2)
arr1
```

```python
arr2 = np.random.randint(1,10,4).reshape(2,2)
arr2
```

```python
common_ele = np.isin(arr1,arr2)
print('Common Element : ',arr1[common_ele])
```

```python
common_ele = np.intersect1d(arr1,arr2)
print('Common Element : ',common_ele)
```

• Subtract row mean from each row of a 2D array

```python
arr = np.random.randint(1,10,4).reshape(2,2)
arr
```

```python
mean_val = np.mean(arr,axis=1,keepdims = True)
mean_val
```

```python
arr - mean_val
```

• Given a 4×4 array, swap the first half rows with the second half rows

```python
arr = np.random.randint(1,25,16).reshape(4,4,)
arr
```

```python
half = np.size(arr[0])//2
swaped = np.vstack((arr[half:],arr[:half]))
swaped
```

• Given a 1D array of random numbers, find the indices of the 5 largest values (no sorting/ check argsort() instead).

```python
arr = np.random.randint(1,50,10)
arr
```

```python
np.argsort(arr)[np.size(arr)-5:]
```

• From a 2D NumPy array, extract only the unique rows (no duplicates hint: check unique method).

```python
arr = np.array([[1,2,3],[4,5,6],[1,2,3]])
arr
```

```python
unique_rows, counts = np.unique(arr, axis=0, return_counts=True)
```

```python
unique_only = unique_rows[counts == 1]
```
