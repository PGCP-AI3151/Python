```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

data = {
    "year" : [2010,2011,2012,2013,2014,2015,2016,2017],
    "india" : [1000,1200,1300,1400,1500,1600,1700,1800],
    "urban" : [200,300,400,326,289,678,459,567]
}

df = pd.DataFrame(data)
df
    
```

```python
plt.plot(df["year"], df["india"], ls = '-', marker = 'o', color = 'blue', label = 'India population')
plt.xlabel("year")
plt.ylabel("population")
plt.title("India population of India")
plt.legend()
plt.show()
```
