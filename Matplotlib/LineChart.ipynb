```python
qqqqaimport pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Year': [2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025],
    'India': [1048, 1167, 1250, 1396, 1408, 1417, 1428, 1441, 1450], # Total population approximation (in millions)
    'Urban': [390, 430, 469, 483, 493, 502, 513, 525, 536]       # Urban population approximation (in millions)
}

df = pd.DataFrame(data)
df
```

```python
plt.plot(df['Year'],df['India'],ls='-', marker = 'o' , color = 'blue',label = 'India Population')
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('India Population Over Year')
plt.legend()
plt.show()
```
