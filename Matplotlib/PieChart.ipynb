```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
count = [9354, 8837, 7456, 5806, 5638]
langs = ['JavaScript', 'HTML/CSS', 'SQL', 'Python', 'Java']
```

```python
plt.pie(count)
```

```python
plt.pie(count, labels = langs, colors = ['red','green','pink','blue','yellow'],
        explode = [0,0,0,0.1,0],
        startangle = 90,
        shadow = True,
        wedgeprops = {'edgecolor' :'black'},
        autopct = '%1.1f%%'
       )
```
