```python
df = pd.read_csv('survey_results-Sample.csv')
df[['Min Age','Max Age']] = df['Age'].str.extract(r'(\d+)-(\d+)').astype(float)
df['Mean Age'] = df[['Min Age','Max Age']].mean(axis=1)
df.dropna()
df
```

```python
bins = [20,30,40,50,60]
plt.hist(df['Mean Age'], bins=bins, edgecolor = 'black')
```
