```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

----- Q1 -----

1.	Draw line chart to show urban and rural population of India over the years using population.
    csv [rural 	population is the difference between total population and urban population]

```python
df = pd.read_csv('Population.csv')
df['Rural Population'] = df['Population'] - df['Urban Population']
to_remove = ['Yearly % Change','Yearly Change','Median Age',
             'Fertility Rate','Density','Urban Pop %','Country\'s Share']
df.drop(to_remove,axis = 1,inplace=True)
df
```

```python
plt.plot(df['Year'],df['Urban Population'],ls='--', color='#058d96', label='Urban Population')
plt.plot(df['Year'],df['Rural Population'], ls='-.', color='#9f33de', label='Rural Population')
plt.xlabel('Year')
plt.ylabel('Population in Billion')
plt.legend()
plt.title('Population Over Year')
plt.show()
```

----- Q2 ------

2. 	Using a Bar chart display India's total, urban and rural population from year 2020 to 2025

```python
df1 = df[13:]
x_values =df1['Year'].values
x_value = np.arange(len(x_values))
width = 0.2
```

```python
plt.bar(x_value - width,df1['Population'],width = width,label = 'Total Population',color = '#8133d4')
plt.bar(x_value,df1['Urban Population'],width = width, label = 'Urban Population',color = '#374151')
plt.bar(x_value + width,df1['Rural Population'],width = width, label = 'Rural Pupolation',color = '#18bacc')
plt.xticks(x_value,x_values)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('Population Over Year')
plt.legend()
plt.show()
```

----- Q3 -----

3.	Using gapminder2007.csv show life expectancy of top 10 highly populated countries from Asia

```python
df = pd.read_csv('gapminder2007.csv')
new_df = df[df['continent'] == 'Asia'].sort_values(by = 'pop',ascending = False).head(10)
new_df
```

```python
plt.bar(new_df['country'],new_df['lifeExp'],color = '#374151')
plt.xticks(rotation = 45)
plt.xlabel('Country')
plt.ylabel('Life Expectancy')
plt.title('County With it\'s Life Expectancy')
plt.show()
```

----- Q4 -----

4.	Using a pie chart show distribution of top six selling cars [Create data as required]

```python
car_data = {
    'Car Model': ['Toyota Corolla', 'Toyota RAV4', 'Tesla Model Y', 'Ford F-Series', 'Honda CR-V', 
                  'Chevrolet Silverado'],
    'Units Sold (Millions)': [1.12, 1.07, 0.98, 0.78, 0.71, 0.62]
}
df = pd.DataFrame(car_data)
df
```

```python
plt.pie(df['Units Sold (Millions)'] , labels = df['Car Model'], 
        colors = ['#1F2937', '#374151', '#4B5563', '#6B7280', '#9CA3AF', '#D1D5DB'],
        startangle = 90,
        explode = [0.15,0,0,0,0,0],
        shadow = True,
        autopct = '%1.2f%%',
        wedgeprops = {'edgecolor' : 'black'},
       )
plt.title('Top Six Salling Cars')
plt.show()
```

5.	Using Histogram show distribution of developers by their experience [use survey_data_sample.csv]

```python
df = pd.read_csv('survey_results-sample.csv')
df.isnull().sum()
```

```python
df = df.dropna(subset=['YearsCode'])
df
```

```python
df['YearsCode'].unique()
```

```python
df['YearsCode'] = df['YearsCode'].replace({'Less than 1 year':'0',
                                           'More than 50 years':'50'})
```

```python
df['YearsCode'].unique()
```

```python
clean_data = pd.to_numeric(df["YearsCode"], errors="coerce")
```

```python
clean_data.isnull().sum()
```

```python
plt.hist(clean_data,edgecolor = 'black',color="#1F2937",alpha=0.8)
plt.xlabel('Year Of Experience')
plt.ylabel('Count Of Employee')
plt.title('Employee Distribution Over Year Of Experiance')
plt.show()
```
